```python
import matplotlib.pyplot as plt
```

```python
top_brands = df['brand'].value_counts().head(10)  #listing count of every brand from top.
print(top_brands)
```

```python
plt.figure(figsize=(10, 5))

top_brands.plot(kind='bar')

plt.title('Top 10 Perfume Brands by Listing Count')
plt.xlabel('Brand')
plt.ylabel('Number of Listings')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
brand_counts = df['brand'].value_counts().head(5)

plt.figure(figsize=(8, 8))

plt.pie(
    brand_counts,
    labels=brand_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Perfume Brands by Listing Count')
plt.show()
```
